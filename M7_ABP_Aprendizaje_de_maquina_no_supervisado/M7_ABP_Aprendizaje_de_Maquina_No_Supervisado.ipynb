## **Lección 1 - *Fundamentos del Aprendizaje No Supervisado***

📌 **Objetivo:** Comprender los fundamentos teóricos del aprendizaje no supervisado y sus diferencias con el aprendizaje supervisado.

📍 **Tareas solicitadas en el ejercicio:**

1. **Identificar las principales tareas que resuelve el aprendizaje no supervisado.**

El aprendizaje no supervisado es una rama del machine learning donde el modelo trabaja con **datos sin etiquetas**. El algoritmo descubre patrones, estructuras o agrupaciones internas en los datos por sí solo.

Las principales tareas son:
- **Clusterización**: agrupar elementos similares.
- **Reducción de dimensionalidad**: reducir la cantidad de variables sin perder información.
- **Análisis de reglas de asociación**: busca reglas o correlaciones dentro de grandes bases de datos.
- **Detección de anomalías**: identificar eventos inusuales o datos que se desvían significativamente del comportamiento general.

Es importante seleccionar la técnica adecuada según el tipo de problema, el objetivo del análisis y la naturaleza de los datos. De esta forma, se garantiza que el modelo no solo sea útil, sino también interpretable y eficiente.

2. **Clasificar técnicas por tipo (clusterización, reducción, asociación).**

Las técnicas se clasifican de la siguiente manera:
  - **Técnicas de Clusterización**:
    - **K-Means**: Agrupamiento basado en centroides
    - **DBSCAN**: Agrupamiento basado en la densidad de los puntos
    - **Clustering Jerárquico (Hierarchical clustering)**: Creación de jerarquías de grupos mediante dendrogramas
    - **Mezcla Gaussiana (GMM)**: Modelado probabilístico de clústeres
  
  - **Técnicas de Reducción de Dimensionalidad**:
    - **PCA (Análisis de Componentes Principales)**: Técnica lineal para maximizar la varianza
    - **t-SNE y UMAP**: Algoritmos no lineales.
      - **t-SNE (t-Distributed Stochastic Neighbor Embedding)**: Proyecta datos de alta dimensión preservando las relaciones de proximidad locales; es decir, asegura que los puntos que están cerca en el espacio original permanezcan cerca en la visualización.
      - **UMAP (Uniform Manifold Approximation and Projection)**: Al igual que t-SNE, busca reducir dimensiones, pero se destaca por ser significativamente más rápido. UMAP logra un mejor equilibrio entre la preservación de la estructura local y la global de los datos.

  - **Técnicas de Asociación**:
    - **Algoritmo Apriori**: Utilizado para extraer reglas de asociación significativas.
    - **FP-Growth**: Identifica conjuntos de elementos frecuentes y genera reglas de co-ocurrencia en grandes bases de datos.

  - **Detección de anomalías**:
    - **Isolation Forest**: Funciona identificando observaciones que se desvían del comportamiento normal.
    - **Autoencoder**: Es una arquitectura de red neuronal que aprende representaciones comprimidas de los datos mediante un proceso de codificación y decodificación.


3. **Analizar casos reales de aplicación.**

- **Clusterización**: segmentación de clientes, análisis de comportamiento, agrupamiento de imágenes, procesamiento de lenguaje natural (NLP - agrupamiento de grandes volúmenes de documentos o noticias por temática), análisis genómico y clasificación de especies.
- **Reducción de dimensionalidad**: Visualización de datos, compresión, procesamiento de lenguaje natural.
- **Análisis de reglas de asociación**:	reglas de co-ocurrencia (ej. Análisis de Carrito de Supermercado: clientes que compran pan también compran mantequilla), sistemas de recomendación.
- **Detección de anomalías:** ciberseguridad, fraude financieros, fallas industriales, mantenimiento predictivo.

## **Lección 2 - *Técnicas de Clusterización***

📌 **Objetivo:** Conocer en profundidad la técnica de clusterización y sus aplicaciones.

📍 **Tareas solicitadas en el ejercicio:**

1. **Definir escenarios donde se aplique clusterización.**

    - Marketing: Segmentación de clientes basada en comportamientos de compra para crear campañas personalizadas.
   - Biología: Agrupamiento de genes o especies con características similares y análisis genómico para investigación médica.
   - Procesamiento de Lenguaje Natural (NLP): Organización de grandes volúmenes de documentos, noticias o comentarios por temáticas sin entrenamiento previo.
   - Ingeniería: Agrupación de sensores con lecturas similares para facilitar la detección de fallas en sistemas industriales.
    - Astronomía: Clasificación de objetos celestes según sus propiedades físicas y espectrales.

2. **Explicar ventajas y desventajas de la técnica.**

**Ventajas:**
- Permite trabajar con grandes volúmenes de datos no estructurados (texto, imágenes).
- Identifica estructuras y relaciones que no son evidentes a simple vista.
- Sirve como paso previo para modelos supervisados, definiendo etiquetas automáticas.

**Desventajas y Limitaciones**
- Validación compleja: Al no existir una "respuesta correcta", validar los resultados requiere métricas adicionales como el índice de Silhouette.
- Sensibilidad: Los resultados dependen críticamente del preprocesamiento (escalado de variables) y del método de distancia elegido.
- Interpretación: El algoritmo no define el significado del grupo; el analista debe traducir la agrupación en información accionable.

3. **Comparar K-Means, DBSCAN y agrupamiento jerárquico teóricamente.**

| Característica | K-Means | DBSCAN | Agrupamiento Jerárquico (Hierarchical Clustering) |
|---|---|---|---|
| Fundamento | Basado en centroides (distancia a puntos medios) | Basado en densidad de puntos en el espacio | Estructura de árbol o dendrograma (aglomerativo/divisivo) |
| Número de Clústeres k | Requiere definirse previamente | No requiere definirse previamente | Se decide después visualizando el dendrograma  |
| Forma de los Clústeres | Esféricos, similares tamaños. | Detecta formas arbitrarias y complejas | Arbitraria, basada en la jerarquía de fusiones |
| Sensibilidad ante Outliers | Muy sensible (distorsionan los centroides) | Robusto (clasifica puntos aislados como ruido) | Sensible a errores iniciales que se propagan en la jerarquía |
| Escalabilidad | Alta (eficiente para grandes datasets) | Media (baja rendimiento en alta dimensionalidad) | Baja (ineficiente en conjuntos de datos muy grandes)|


4. **Asociar cada algoritmo a un tipo de dato o problema real**

- **K-Mean**: Ideal para compresión de imágenes (reduciendo el número de colores) o segmentación de mercado donde se busca una división clara y balanceada de grupos.
- **DBSCAN**: El más adecuado para la detección de anomalías y fraudes, donde es crucial separar el comportamiento normal de puntos aislados (ruido).
- **Agrupamiento Jerárquico**: Preferido en biología y taxonomía, ya que permite visualizar relaciones evolutivas o funcionales complejas a través de su estructura multinivel.

### Preprocesamiento del Dataset

```python
# Copiar Dataset Original y trabajar en la copia
df = df_raw.copy()
```

```python
# Eliminar columnas innecesarias
# 'ID' no aporta
# 'Segmentation' indica los segmentos actuales de los clientes.
df = df.drop(columns=['ID', 'Segmentation'])
df.head(3)
```

```python
# Definir tipos de columnas:
# Columnas Númericas
numeric = ['Age', 'Work_Experience', 'Family_Size']

# Columnas Categórica Ordinal: 'Spending_Score' tiene orden
ordinal = ['Spending_Score']
ordinal_categories = [['Low', 'Average', 'High']]

# Categóricas Nominales
nominal = ['Gender', 'Ever_Married', 'Graduated', 'Profession', 'Var_1']
```

```python
# Imputación de Valores Nulos

# Imputar los valores faltantes en las columnas numéricas con la Mediana (robusta a outliers)
imputer_num = SimpleImputer(strategy='median')
df[numeric] = imputer_num.fit_transform(df[numeric])

# Imputar los valores faltantes en columnas categóricas con la Moda
imputer_cat = SimpleImputer(strategy='most_frequent')
df[ordinal + nominal] = imputer_cat.fit_transform(df[ordinal + nominal])

# Verificar si ya no hay valores nulos
df.isnull().sum()
```

```python
# Identificar Outliers - Método del rango intercuartil (IQR)

# Identificar el primer, tercer cuartil y el IQR para cada columna numérica
Q1 = df[numeric].quantile(0.25)
Q3 = df[numeric].quantile(0.75)
IQR = Q3 - Q1

# Calculo de límites para detectar outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Detectar outliers
outliers = (df[numeric] < limite_inferior) | (df[numeric] > limite_superior)

# Contar los outliers por columna
outliers_count = outliers.sum()
display(outliers_count)
```

```python
print(f'Total de outliers: {outliers_count.sum()}')
```

```python
# Boxplots
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.suptitle('Detección de Outliers — Boxplots', fontweight='bold')

for ax, col in zip(axes, numeric):
    ax.boxplot(df[col].dropna())
    ax.set_title(col)
    ax.set_ylabel('Valor')

plt.tight_layout()
plt.savefig('02_outliers.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Eliminar Outliers
mask = ((df[numeric] >= limite_inferior) & (df[numeric] <= limite_superior)).all(axis=1)

filas_antes = len(df)
df = df[mask].reset_index(drop=True)
filas_despues = len(df)

print(f'N° de Filas originales:   {filas_antes}')
print(f'N° de Filas después:      {filas_despues}')
print(f'Outliers eliminados:      {filas_antes - filas_despues} ({(filas_antes - filas_despues)/filas_antes*100:.1f}%)')
```

```python
# Pipeline de Escalamiento y Codificación
# Variables Númericas
numeric_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())  # nulos ya imputados, solo escalar (estandarizar los datos con media=0, std=1)
])

# Variable Categórica Ordinal: 'Spending_Score'
ordinal_transformer = Pipeline(steps=[
    ('encoder', OrdinalEncoder(categories=ordinal_categories))
])

# Variables Nominales
nominal_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# Unir todo
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric),
    ('ord', ordinal_transformer, ordinal),
    ('nom', nominal_transformer, nominal)
])

# Ajuste y transformación
X_elaborado = preprocessor.fit_transform(df)
print(f'Dimensiones: {X_elaborado.shape}')
X_elaborado
```

### 2. **Implementar PCA y t-SNE con scikit-learn.**

#### Reducción de dimensionalidad: PCA — Análisis de Componentes Principales

```python
# Crear el modelo y reducir a 2 componentes para visualización
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_elaborado)

# Varianza explicada por cada componente
print('Varianza Explicada:')
print(f'PC1: {pca.explained_variance_ratio_[0]:.2%}')
print(f'PC2: {pca.explained_variance_ratio_[1]:.2%}')
print(f'Total: {pca.explained_variance_ratio_.sum():.2%}')
```

El gráfico muestra una nube densa sin agrupaciones visibles.

## **Lección 4 - *Aplicación práctica de Clusterización con Python***

📌 **Objetivo:** Implementar un pipeline de clusterización aplicando K-Means, DBSCAN y agrupamiento jerárquico.

📍 **Tareas solicitadas en el ejercicio:**
- Aplicar los tres algoritmos sobre el dataset preprocesado
- Visualizar los resultados en 2D tras reducción
- Calcular coeficiente de silueta y método del codo
- Determinar el número óptimo de clústeres

#### **Algoritmo K-Means**

```python
# Selección del Valor K: Método del Codo
X_clust = X_elaborado

inercias = []

K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    km.fit(X_clust)
    inercias.append(km.inertia_)

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(K_range, inercias, 'bo-', linewidth=2, markersize=8)
plt.title('Método del Codo')
plt.xlabel('Número de Clústeres (k)')
plt.ylabel('Inercia (WCSS)')
plt.xticks(K_range)
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.savefig('06_metodo_codo.png', dpi=150, bbox_inches='tight')
plt.show()
```

De acuerdo con el Método del Codo, el K óptimo es k = 5.

Se analizará el coeficiente de silueta para ver si indica el mismo k.

```python
# Coeficiente de Silueta para diferentes k

siluetas = []

for k in K_range:
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_clust)
    siluetas.append(silhouette_score(X_clust, labels))

# K óptimo
K_OPTIMO = list(K_range)[siluetas.index(max(siluetas))]
print(f'k óptimo según Silueta: k = {K_OPTIMO}')
print(f'Silueta máxima con  k = {K_OPTIMO} : {max(siluetas):.4f}')
```

```python
# Entrenar K-Means con el k elegido en el método del codo
K_codo = 5  # valor identificado en el gráfico del codo
km = KMeans(n_clusters=K_codo, random_state=RANDOM_STATE, n_init=10)
km.fit(X_clust)

# Calcular coeficiente de silueta
score = silhouette_score(X_clust, km.labels_)
print(f'k según Método del Codo: k = {K_codo}')
print(f'Coeficiente de Silueta con k = {K_codo} : {score:.4f}')
```

Hay una discrepancia entre ambos métodos, lo cual es común cuando los datos no tienen clusters perfectamente definidos. Las opciones son:

- Con k=3 se genera una mejor calidad de clustering según silueta. Grupos más compactos y separados, aunque sean menos.
- Con k=5, sugerido por el codo, se generan más segmentos. Esto permite estrategias de marketing más específicas.

Por lo que, a pesar que k=3 es valor de silueta recomendado, en este caso, dado que la diferente entre los coeficiente es pequeña y el objetivo del análisis es segmentar clientes se elegirá k=5.

```python
# K-MEANS
km_final = KMeans(n_clusters=K_codo, random_state=RANDOM_STATE, n_init=10)
km_final.fit(X_clust)
labels_km = km_final.labels_

# Tamaño de cada clúster
print('Distribución de clústeres')
for i in range(K_codo):
    n = (labels_km == i).sum()
    print(f'  Clúster {i}: {n} clientes ({n/len(labels_km)*100:.1f}%)')

# Agregar clúster al DataFrame original
df['Cluster_KMeans'] = labels_km

# Silueta
score_km = silhouette_score(X_clust, labels_km)
print(f'\nCoeficiente de Silueta K-Means (k={K_codo}): {score_km:.4f}')
```

```python
# Visualización sobre PCA
pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
pca_df['Cluster'] = labels_km

plt.figure(figsize=(10, 8))
sns.scatterplot(x='PC1', y='PC2', hue='Cluster', data=pca_df, palette='tab10', s=60, alpha=0.8)
plt.title(f'K-Means (k={K_codo}) — Vista PCA', fontsize=14)
plt.xlabel('PC 1')
plt.ylabel('PC 2')
plt.legend(title='Clúster', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.savefig('07_kmeans.png', dpi=150, bbox_inches='tight')
plt.show()
```

El clúster 3 (rojo) y el clúster 0 (azul) son los más separados del resto.

El solapamiento central entre naranja, verde y púrpura refleja que estos tres perfiles comparten características similares en las dimensiones capturadas por PCA.

El solapamiento es consistente con la silueta de 0.1660, ya que los grupos existen, pero con fronteras difusas.

#### **Algoritmo DBSCAN**

```python
# Probar distintos valores de eps en DBSCAN
for eps_val in [1.0, 1.2, 1.4, 1.45, 1.46, 1.5, 2.0, 3.0]:
    dbs_test = DBSCAN(eps=eps_val, min_samples=5)
    labels_test = dbs_test.fit_predict(X_clust)
    n_cl = len(set(labels_test)) - (1 if -1 in labels_test else 0)
    n_ru = list(labels_test).count(-1)
    print(f'eps={eps_val} → Clústeres: {n_cl} | Ruido: {n_ru} ({n_ru/len(labels_test)*100:.1f}%)')
```

El salto brusco que ocurre entre eps=1.4 (160 clústeres) y eps=1.45 (2 clústeres) confirma el problema de la maldición de la dimensionalidad. Como el dataset tiene 26 dimensiones, no hay una zona de transición suave como en datasets pequeños. Esto es una desventaja de este algoritmo.

```python
# DBSCAN eps=1.4
dbscan14 = DBSCAN(eps=1.4, min_samples=5)
labels_dbs14 = dbscan14.fit_predict(X_clust)

# Resultados
n_clusters_dbs14 = len(set(labels_dbs14)) - (1 if -1 in labels_dbs14 else 0)
n_noise14 = list(labels_dbs14).count(-1)
print(f'Clústeres encontrados: {n_clusters_dbs14}')
print(f'Puntos de ruido: {n_noise14} ({n_noise14/len(labels_dbs14)*100:.1f}%)')

# Silueta (excluyendo ruido)
X_clustered14 = X_clust[labels_dbs14 != -1]
labels_clustered14 = labels_dbs14[labels_dbs14 != -1]
if len(set(labels_clustered14)) > 1:
    score_dbs14 = silhouette_score(X_clustered14, labels_clustered14)
    print(f'Coeficiente de Silueta DBSCAN eps=1.4 (sin ruido): {score_dbs14:.4f}')
```

```python
# DBSCAN FINAL/ELEGIDO eps=1.45
dbscan145 = DBSCAN(eps=1.45, min_samples=5)
labels_dbs145 = dbscan145.fit_predict(X_clust)

# Resultados
n_clusters_dbs145 = len(set(labels_dbs145)) - (1 if -1 in labels_dbs145 else 0)
n_noise145 = list(labels_dbs145).count(-1)
print(f'Clústeres encontrados: {n_clusters_dbs145}')
print(f'Puntos de ruido: {n_noise145} ({n_noise145/len(labels_dbs145)*100:.1f}%)')

# Silueta (excluyendo ruido)
X_clustered145 = X_clust[labels_dbs145 != -1]
labels_clustered145 = labels_dbs145[labels_dbs145 != -1]
if len(set(labels_clustered145)) > 1:
    score_dbs145 = silhouette_score(X_clustered145, labels_clustered145)
    print(f'Coeficiente de Silueta DBSCAN eps=1.45 (sin ruido): {score_dbs145:.4f}')
```

```python
# Comparación visual DBSCAN: eps=1.4 vs eps=1.45
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('DBSCAN', fontsize=14, fontweight='bold')

for ax, labels, titulo in zip(axes,
    [labels_dbs14, labels_dbs145],
    [f'eps=1.4  | {n_clusters_dbs14} clústeres | Ruido: {n_noise14} ({n_noise14/len(labels_dbs14)*100:.1f}%) | Silueta: {score_dbs14:.4f}',
     f'eps=1.45 | {n_clusters_dbs145} clústeres | Ruido: {n_noise145} ({n_noise145/len(labels_dbs145)*100:.1f}%) | Silueta: {score_dbs145:.4f}']):

    df_plot = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
    df_plot['Cluster'] = labels.astype(str)
    sns.scatterplot(x='PC1', y='PC2', hue='Cluster',
                    data=df_plot, palette='tab10',
                    s=40, alpha=0.7, ax=ax, legend=False)
    ax.set_title(titulo, fontsize=9)
    ax.set_xlabel('PC 1')
    ax.set_ylabel('PC 2')
    ax.grid(True, alpha=0.4)

plt.tight_layout()
plt.savefig('08_dbscan_comparacion.png', dpi=150, bbox_inches='tight')
plt.show()
```

DBSCAN no es el algoritmo más adecuado para este dataset, porque la alta dimensionalidad impide identificar diferencias de densidad significativas. Su aporte principal es la detección de 484 clientes atípicos.

#### **Algoritmo Agrupamiento Jerárquico (Hierarchical Clustering)**

```python
# Dendrograma: explorar estructura y decidir k
Z = linkage(X_clust, method='ward')

plt.figure(figsize=(14, 6))
dendrogram(Z, truncate_mode='lastp', p=20, color_threshold=60)
plt.axhline(y=60, color='r', linestyle='--', label='Línea de corte')
plt.title('Dendrograma — Agrupamiento Jerárquico (Ward)')
plt.xlabel('Observaciones')
plt.ylabel('Distancia')
plt.legend()
plt.tight_layout()
plt.savefig('09_dendrograma.png', dpi=150, bbox_inches='tight')
plt.show()
```

El corte del dendrograma en y=60 produce 5 clústeres, confirmando el mismo k sugerido por el método del codo.

```python
from scipy.cluster.hierarchy import fcluster

# Cortar en la misma altura que la línea roja del dendrograma
labels_fcluster = fcluster(Z, t=60, criterion='distance')

n_clusters_fc = len(set(labels_fcluster))
print(f'Clústeres con corte en t=60: {n_clusters_fc}')
print(f'Distribución:')
for c in sorted(set(labels_fcluster)):
    n = (labels_fcluster == c).sum()
    print(f'  Clúster {c}: {n} clientes ({n/len(labels_fcluster)*100:.1f}%)')
```

```python
# Aplicar AgglomerativeClustering con k=5
from sklearn.cluster import AgglomerativeClustering

modelo_hier5 = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')
labels_hier5 = modelo_hier5.fit_predict(X_clust)

# Distribución de clústeres
print('Distribución de clústeres con k=5')
cluster_size5 = pd.Series(labels_hier5).value_counts().sort_index()
cluster_pct5  = pd.Series(labels_hier5).value_counts(normalize=True).sort_index() * 100
for i in cluster_size5.index:
    print(f'  Clúster {i}: {cluster_size5[i]} clientes ({cluster_pct5[i]:.1f}%)')

# Coeficiente de Silueta
score_hier5 = silhouette_score(X_clust, labels_hier5)
print(f'\nCoeficiente de Silueta Jerárquico (k=5): {score_hier5:.4f}')
```

```python
# Silueta para distintos k
print('Silueta por número de clústeres:')
for k in range(2, 8):
    model = AgglomerativeClustering(n_clusters=k)
    labels = model.fit_predict(X_clust)
    score = silhouette_score(X_clust, labels)
    print(f'k={k}  →  Silueta: {score:.4f}')
```

El máximo está en k=3 con 0.1804, igual que en K-Means Óptimo. Por lo que,  en dos algoritmos distintos, K-Means y Jerárquico, indica que utilizar un k = 3 produce grupos de mejor calidad estadística. Pero k = 5 tiene el respaldo de tres fuentes distintas: método del codo (k=5), dendograma (k=5) y los segmentos originales del dataset (k=4).

Adicionalmente, la diferencia de silueta entre k=3 y k=5 es 0.0557 (pequeña). Con una diferencia de 0.0557, la pérdida de calidad estadística se justifica.

Finalmente, al tener cinco perfiles de clientes (k=5) esto permite estrategias de marketing más granulares y diferenciadas.
