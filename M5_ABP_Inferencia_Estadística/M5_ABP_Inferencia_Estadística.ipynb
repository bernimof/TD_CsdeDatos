4. **Calcular probabilidades básicas de eventos (intersección, unión, complementarios)**

- P(A): Probabilidad de que un estudiante tenga menos de 7 horas de sueño.






```python
P_A = (data['Horas_Sueno'] < 7).mean()
print('La probabilidad de que un estudiante tenga menos de 7 horas de sueño es de', P_A*100,'%')
```

-	P(A ∪ B): Probabilidad de que un estudiante tenga menos de 7 horas de sueño o cumpla con los 30 minutos de actividad física (unión).

```python
P_A_union_B = ((data['Horas_Sueno'] < 7) | (data['Cumple_30min_AF'] == 1)).mean()
print('La probabilidad de que un estudiante tenga menos de 7 horas de sueño o \ncumpla con los 30 minutos de actividad física :', P_A_union_B*100,'%')
```

-	P(A’): Probabilidad complementaria de que un estudiante tenga más de 7 horas de sueño.

## **Lección 3: *Distribución de probabilidad***

📌 **Objetivo:** Aplicar distribuciones adecuadas según el tipo de variable y
contexto.

📍 **Tareas desarrolladas:**

1. **Identificar si las variables se ajustan a distribuciones discretas (binomial, Poisson) o continuas (normal).**

Las variables cuantitativas 'Horas_Sueño' y 'Min_Actividad_Fisica' son continuas, por lo que se ajustarían a distribuciones como la normal.

La variable 'Cumple_30min_AF' es binaria discreta, por lo tanto, se ajusta a una distribución binomial si se analiza como la cantidad de éxitos (cumplir con la actividad física) en una serie de intentos.

2. **Justificar la elección de las funciones de distribución.**

-	Distribución Normal para 'Horas_Sueño' y 'Min_Actividad_Fisica':

La distribución normal o gaussiana es adecuada para variables continuas, especialmente cuando la muestra es grande y los datos parecen distribuidos de manera simétrica alrededor de la media. De todas maneras, se realizó el test de normalidad de Shapiro-Wilk.

```python
# Test de Normalidad (Shapiro-Wilk) para 'Horas_Sueño' y 'Min_Actividad_Fisica'
shapiro_horas_sueño = stats.shapiro(data['Horas_Sueno'])
print("Prueba de Normalidad ('Horas_Sueño'):")
print("Estadística:", shapiro_horas_sueño.statistic)
print("Valor p:", shapiro_horas_sueño.pvalue)

shapiro_actividad_fisica = stats.shapiro(data['Min_Actividad_Fisica'])
print("\nPrueba de Normalidad ('Min_Actividad_Fisica'):")
print("Estadística:", shapiro_actividad_fisica.statistic)
print("Valor p:", shapiro_actividad_fisica.pvalue)
```

Por lo tanto, al revisar los resultados obtenidos, se puede concluir que:
- La estadística de la prueba Shapiro-Wilk mide cuán cerca están los datos de una distribución normal. Cuanto más cercano a 1 sea el valor, más cerca están los datos de una distribución normal. En ambos casos se obtuvieron valores muy cercanos a 1, 'Horas_Sueno' con 0.994 y 'Min_Actividad_Fisica' con 0.986, por lo que según este estadístico, se ajustan bien a una distribución normal.

- El valor p es el resultado de la prueba de normalidad. Un valor p mayor a 0.05 indica que no hay evidencia suficiente para rechazar la hipótesis nula, que afirma que los datos provienen de una distribución normal. En el caso de 'Horas_Sueno', 0.60 es mayor a 0.05, lo que significa que los datos se ajustan bien a una distribución normal. Pero en el caso de 'Min_Actividad_Fisica', el valor p es 0.053, que es ligeramente mayor que 0.05. Esto sugiere que los datos podrían ajustarse a una distribución normal, pero el valor p está justo en el límite, por lo que hay una ligera duda de si se ajustan completamente.

-	Distribución Binomial para 'Cumple_30min_AF':

La distribución binomial surge al repetir un experimento de Bernoulli de manera independiente n veces.

La variable 'Cumple_30min_AF' es cualitativa discreta y binaria, lo que significa que solo tiene dos resultados posibles: éxito (1 = cumple con los 30 min) o fracaso (0 = no cumple). Cada encuesta realizada a un estudiante se considera un ensayo de Bernoulli independiente. El experimento consiste en repetir estos ensayos de manera independiente 200 veces (n=200), que es el tamaño total de la muestra analizada. Por lo que, la distribución binomial es la distribución ideal para esta variable.

3. **Calcular probabilidades con fórmulas o software para las variables clave.**

- Para 'Horas_Sueno' y 'Min_Actividad_Fisica' con distribución normal

```python
# Definir umbrales de valores para los cálculos
umbral_sueño = 7  # Probabilidad de que un estudiante duerma ≤ 7 horas
umbral_actividad = 30  # Probabilidad de que un estudiante haga ≤ 30 minutos de actividad física

# Calculo de la media y la desviación estándar
media_horas_sueño = data['Horas_Sueno'].mean()
ds_horas_sueño = data['Horas_Sueno'].std()

media_actividad_fisica = data['Min_Actividad_Fisica'].mean()
ds_actividad_fisica = data['Min_Actividad_Fisica'].std()

# Calculo de probabilidades con la distribución normal para 'Horas_Sueno'
P_sueño_menor_7 = stats.norm.cdf(umbral_sueño, media_horas_sueño, ds_horas_sueño)  # P(X ≤ 7)
P_sueño_mayor_7 = 1 - P_sueño_menor_7  # P(X > 7)

# Calculo de probabilidades con la distribución normal para 'Min_Actividad_Fisica'
P_actividad_menor_30 = stats.norm.cdf(umbral_actividad, media_actividad_fisica, ds_actividad_fisica)  # P(X ≤ 30)
P_actividad_mayor_30 = 1 - P_actividad_menor_30  # P(X > 30)

# Mostrar resultados
print(f"Probabilidad de dormir ≤ 7 horas: {P_sueño_menor_7:.4f}")
print(f"Probabilidad de dormir > 7 horas: {P_sueño_mayor_7:.4f}")
print(f"\nProbabilidad de hacer ≤ 30 minutos de actividad física: {P_actividad_menor_30:.4f}")
print(f"Probabilidad de hacer > 30 minutos de actividad física: {P_actividad_mayor_30:.4f}")
```

- Para 'Cumple_30min_AF' con distribución binomial

```python
# Número de estudiantes
n_estudiantes = len(data)

# Probabilidad de éxito (cumple con 30 minutos de actividad física)
p_cumple_30min = data['Cumple_30min_AF'].mean()

# Calculo de probabilidad de exactamente k estudiantes cumpliendo la actividad física
k = 100
binomial_prob = stats.binom.pmf(k, n_estudiantes, p_cumple_30min)

# Mostrar resultado
print(f"La probabilidad de que exactamente {k} estudiantes cumplan \ncon los 30 minutos de actividad física es: {binomial_prob:.4f}")
```

## **Lección 4: Distribución muestral y teorema del límite central**

📌 **Objetivo:** Aplicar el concepto de distribución muestral e interpretar el teorema del límite central (TLC).

**📍 Tareas desarrolladas:**

1. **Generar distribuciones muestrales de media a partir del dataset simulado**

```python
# Define tamaño de la muestra y número de repeticiones
n_muestras = 1000  # Número de muestras a tomar
tamaño_muestra = 30  # Tamaño de cada muestra

# Genera distribuciones muestrales para 'Horas_Sueño'
medias_horas_sueño = []
for _ in range(n_muestras):
    muestra = data['Horas_Sueno'].sample(n=tamaño_muestra, replace=True)
    medias_horas_sueño.append(muestra.mean())

# Genera distribuciones muestrales para 'Min_Actividad_Fisica'
medias_actividad_fisica = []
for _ in range(n_muestras):
    muestra = data['Min_Actividad_Fisica'].sample(n=tamaño_muestra, replace=True)
    medias_actividad_fisica.append(muestra.mean())
```

2. **Verificar empíricamente el Teorema del Límite Central**

El Teorema del Límite Central (TLC) indica que a medida que tomamos más y más muestras y calculamos sus medias, esas medias muestrales se aproximan a una distribución normal, independientemente de la distribución original de los datos.

Por lo que, a continuación, se graficarán las distribuciones muestrales de la media, calculadas anteriormente, para ver si tienen forma de campana (aproximándose a una distribución normal). Además, están los histogramas originales, para facilitar la comparación de las gráficas.

```python
# Verificación empíricamente del Teorema del Límite Central (TLC)
plt.figure(figsize=(14, 12))

# Primera fila: Distribuciones Muestrales de la Media
# Distribución muestral de la media para 'Horas_Sueno'
plt.subplot(2, 2, 1)
sns.histplot(medias_horas_sueño, kde=True, color='skyblue')
plt.title('Distribución Muestral de la Media Horas de Sueño')
plt.xlabel('Media de Horas de Sueño')
plt.ylabel('Densidad')

# Distribución muestral de la media para 'Min_Actividad_Fisica'
plt.subplot(2, 2, 2)
sns.histplot(medias_actividad_fisica, kde=True, color='salmon')
plt.title('Distribución Muestral de la Media Actividad Física')
plt.xlabel('Media de Minutos de Actividad Física')
plt.ylabel('Densidad')

# Segunda fila: Histogramas Originales
# Histograma de 'Horas_Sueno’
plt.subplot(2, 2, 3)
sns.histplot(data['Horas_Sueno'], bins=15, kde=True, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Distribución de Horas de Sueño')
plt.xlabel('Horas de Sueño')
plt.ylabel('Densidad')

# Histograma de 'Min_Actividad_Fisica'
plt.subplot(2, 2, 4)
sns.histplot(data['Min_Actividad_Fisica'], bins=15, kde=True, color='salmon', edgecolor='black', alpha=0.7)
plt.title('Distribución de Minutos de Actividad Física')
plt.xlabel('Minutos de Actividad Física')
plt.ylabel('Densidad')

plt.tight_layout()
plt.show()
```

A través de estos gráficos, podemos verificar empíricamente el Teorema del Límite Central (TLC), ya que independientemente de la forma de los histogramas de los datos originales, las medias de muestras grandes tienden a seguir una distribución normal.

## **Lección 5: Inferencia e intervalos de confianza para la media**

📌 **Objetivo:** Estimar parámetros poblacionales mediante intervalos de confianza.

📍 **Tareas desarrolladas:**

1. **Calcular intervalos de confianza para la media de al menos dos variables.**

```python
# Nivel de confianza (95%)
confianza = 0.95

# Cálculo de intervalo de confianza para la media (t de Student) de 'Horas_Sueno’
t_horas_sueño = stats.t.ppf((1 + confianza) / 2, len(data) - 1)
margen_error_horas_sueño = t_horas_sueño * (ds_horas_sueño / np.sqrt(len(data)))

# Intervalo de confianza para 'Horas_Sueno’
ic_horas_sueño = (media_horas_sueño - margen_error_horas_sueño, media_horas_sueño + margen_error_horas_sueño)

# Cálculo de intervalo de confianza para la media (t de Student) de 'Min_Actividad_Fisica'
t_actividad_fisica = stats.t.ppf((1 + confianza) / 2, len(data) - 1)
margen_error_actividad_fisica = t_actividad_fisica * (ds_actividad_fisica / np.sqrt(len(data)))

# Intervalo de confianza para 'Min_Actividad_Fisica'
ic_actividad_fisica = (media_actividad_fisica - margen_error_actividad_fisica, media_actividad_fisica + margen_error_actividad_fisica)

print("\nIC 95% 'Horas_Sueno': ",ic_horas_sueño)
print("\nIC 95% 'Min_Actividad_Fisica': ",ic_actividad_fisica)
```

Este cálculo significa que la media poblacional de horas de sueño está entre 6.25 y 6.57 horas con un 95% de certeza. Y la media poblacional de minutos de actividad física está entre 27.11 y 29.95 minutos con un 95% de certeza.

2. **Interpretar los resultados con distintos niveles de confianza (90%, 95%, 99%).**

```python
# Nivel de confianza (90%)
confianza = 0.90

# Cálculo de intervalo de confianza para la media (t de Student) de 'Horas_Sueno’
t_horas_sueño = stats.t.ppf((1 + confianza) / 2, len(data) - 1)
margen_error_horas_sueño = t_horas_sueño * (ds_horas_sueño / np.sqrt(len(data)))

# Intervalo de confianza para 'Horas_Sueno’
ic_horas_sueño = (media_horas_sueño - margen_error_horas_sueño, media_horas_sueño + margen_error_horas_sueño)

# Cálculo de intervalo de confianza para la media (t de Student) de 'Min_Actividad_Fisica'
t_actividad_fisica = stats.t.ppf((1 + confianza) / 2, len(data) - 1)
margen_error_actividad_fisica = t_actividad_fisica * (ds_actividad_fisica / np.sqrt(len(data)))

# Intervalo de confianza para 'Min_Actividad_Fisica'
ic_actividad_fisica = (media_actividad_fisica - margen_error_actividad_fisica, media_actividad_fisica + margen_error_actividad_fisica)

print("\nIC 90% 'Horas_Sueno': ",ic_horas_sueño)
print("\nIC 90% 'Min_Actividad_Fisica': ",ic_actividad_fisica)
```

📌 **Objetivo:** Realizar pruebas de hipótesis para validar afirmaciones sobre la población.

📍 **Tareas desarrolladas:**

2. **Calcular valor-p y compararlo con α para tomar decisiones.**


***Queremos evaluar déficit de sueño. Esto ocurre cuando se duerme menos de 7 horas.***

* H₀ : La media de horas de sueño es igual a 7 horas (μ = 7).
* H₁ : La media de horas de sueño es menor a 7 horas (μ < 7).



```python
# Test de hipótesis para la media de Horas de Sueño (evaluar déficit de sueño)
# H₀: μ = 7, H₁: μ < 7

# Test t de Student (una cola)
t_horas_sueño = (media_horas_sueño - 7) / (ds_horas_sueño / np.sqrt(len(data)))
df_horas_sueño = len(data) - 1
p_horas_sueño = stats.t.cdf(t_horas_sueño, df_horas_sueño)  # Valor p para la prueba unilateral

# Resultados de las pruebas de hipótesis
t_horas_sueño, p_horas_sueño
```

El valor p es de 1.48 × 10⁻¹², o sea, muy pequeño (mucho menor que α = 0.05 [un 5% de error]), lo que significa que rechazamos la hipótesis nula (H₀: μ = 7).

Por lo tanto, hay evidencia suficiente para afirmar que la media de horas de sueño es menor a 7 horas en la población.

***Queremos evaluar si al menos el 50% cumple con ≥30 minutos diarios de actividad física.***

  * H₀: La proporción de estudiantes que cumple con ≥30 minutos diarios de actividad física es igual a 50% (p = 0.50).
  * H₁: La proporción de estudiantes que cumple con ≥30 minutos diarios de actividad física es menor a 50% (p < 0.50).

```python
# Test de hipótesis para la proporción de estudiantes que cumplen con ≥30 minutos diarios de actividad física
# H₀: p = 0.50, H₁: p < 0.50

# Proporción de estudiantes que cumplen con ≥30 minutos de actividad física
p_cumple_30min = data['Cumple_30min_AF'].mean()

# Test de proporciones (una cola)
z_actividad_fisica = (p_cumple_30min - 0.50) / np.sqrt((0.50 * (1 - 0.50)) / len(data))
p_actividad_fisica = stats.norm.cdf(z_actividad_fisica)  # Valor p para la prueba unilateral

# Resultados de las pruebas de hipótesis
z_actividad_fisica, p_actividad_fisica
```

El valor p es 0.839, que es mucho mayor que 0.05.
Por lo tanto, no hay suficiente evidencia para rechazar la hipótesis nula (H₀: p = 0.50). Esto significa que no podemos afirmar que menos del 50% de los estudiantes cumplen con los 30 minutos de actividad física diaria.

3. **Explicar errores tipo I y II en contexto del estudio.**

En estadística, cuando realizamos una prueba de hipótesis, existe la posibilidad de cometer errores al tomar decisiones basadas en los resultados. Los dos tipos de errores más comunes son:

1. ***Error tipo I (α)***: Probabilidad de rechazar la hipótesis nula (H₀) cuando es verdadera (falso positivo).
    - El error Tipo I ocurre cuando afirmamos que hay un efecto o una diferencia, cuando en realidad no la hay.

    - Este error está relacionado con el nivel de significancia de la prueba (α). Si α = 0.05, el riesgo de cometer un error Tipo I es del 5%.

    - Si cometemos un error Tipo I en el test para déficit de sueño, esto significaría que rechazamos incorrectamente que la media es igual a 7 horas (H₀: μ = 7), cuando en realidad la media de horas de sueño realmente es 7 horas. Esto podría hacer que afirmemos que hay un déficit de sueño cuando no existe.

2. ***Error tipo II (β)***: Probabilidad de no rechazar la hipótesis nula (H₀) cuando es falsa (falso negativo).

   - El error Tipo II ocurre cuando no detectamos un efecto o una diferencia que realmente existe.
    
    - El riesgo de cometer un error Tipo II está relacionado con la potencia de la prueba (1 - β). La potencia es la probabilidad de rechazar correctamente la hipótesis nula cuando es falsa.

    - Si cometemos un error Tipo II en el test para déficit de sueño, esto significaría que no rechazamos la hipótesis de que la media es igual a 7 horas (H₀: μ = 7), cuando en realidad la media de horas de sueño es menor a 7 horas. Es decir, concluimos que no hay déficit de sueño cuando realmente sí lo hay.
